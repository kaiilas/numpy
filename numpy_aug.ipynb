```python
z=np.array([[1.6,2.5,3.6,4.8],[5.1,6.1,7.1,8.1],[1.2,2.2,3.2,4.2]])
print(z.max())
print(z.min())
print(z.sum())
```

```python
z=np.array([[1.6,2.5,3.6,4.8],[5.1,6.1,7.1,8.1],[1.2,2.2,3.2,4.2]])
print(z.max(axis=1))
print(z.max(axis=0))
print(np.sqrt(z))
print(np.sin(z))
print(np.cos(z))
```
